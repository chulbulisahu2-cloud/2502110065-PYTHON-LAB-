```python
class CentrifugalPump:
    def __init__(self, name: str, max_flow_rate: float):
        self.name = name
        self.__max_flow_rate = max_flow_rate  # Private rating limit
        self.__current_flow = 0.0              # Private state
        self.__is_running = False             # Private state

    def __str__(self):
        return f"Pump Nameplate: {self.name} [Max Flow: {self.__max_flow_rate} L/min]"

    def set_flow_rate(self, flow: float):
        """Validating operating method refusing out-of-range commands."""
        if flow < 0:
            print(f"[{self.name} ERROR] Refused: Flow rate cannot be negative.")
            return
        if flow > self.__max_flow_rate:
            print(f"[{self.name} ERROR] Refused: {flow} L/min exceeds max limit of {self.__max_flow_rate} L/min.")
            return

        self.__current_flow = flow
        self.__is_running = flow > 0
        print(f"[{self.name}] Flow rate set to {self.__current_flow} L/min.")

    def status(self) -> str:
        """Getter for machine status."""
        state = "RUNNING" if self.__is_running else "IDLE"
        return f"[{self.name}] Status: {state} | Current Flow: {self.__current_flow}/{self.__max_flow_rate} L/min"

# Demonstration
pump = CentrifugalPump("Pump-101", 200.0)
print(pump.status())

# Out of range test
pump.set_flow_rate(250.0)  # Refused
pump.set_flow_rate(150.0)  # Accepted
print(pump.status())

# Note for Lab Record:
# Line to bypass guard: pump._CentrifugalPump__current_flow = 999.0
# Reason allowed: Python uses name mangling rather than hard access restriction to allow debugging and inspection.
```

```python
import random

class CentrifugalPump:
    def __init__(self, name: str, max_flow_rate: float):
        self.name = name
        self.__max_flow_rate = max_flow_rate
        self.__total_volume_pumped = 0.0
        self.__jobs_processed = 0
        self.__jobs_rejected = 0

    def process_job(self, requested_flow: float, duration_mins: float):
        if requested_flow <= 0 or requested_flow > self.__max_flow_rate:
            self.__jobs_rejected += 1
            print(f"[{self.name}] Job REJECTED: Requested flow {requested_flow} L/min out of range.")
            return False

        volume = requested_flow * duration_mins
        self.__total_volume_pumped += volume
        self.__jobs_processed += 1
        print(f"[{self.name}] Job SUCCESS: Pumped {volume:.2f} Liters.")
        return True

    def status(self) -> str:
        return (
            f"\n--- SHIFT REPORT: {self.name} ---\n"
            f"Jobs Processed Successfully : {self.__jobs_processed}\n"
            f"Jobs Rejected               : {self.__jobs_rejected}\n"
            f"Total Volume Pumped        : {self.__total_volume_pumped:.2f} L\n"
            f"---------------------------------"
        )

# Run a batch of 10 jobs
pump = CentrifugalPump("Pump-201", 100.0)
random.seed(42)  # For reproducible random output

print("--- STARTING BATCH ---")
for i in range(1, 11):
    req_flow = round(random.uniform(20.0, 130.0), 1) # Some will exceed max limit (100)
    duration = round(random.uniform(1.0, 5.0), 1)
    print(f"Job #{i}: Flow={req_flow} L/min for {duration} min -> ", end="")
    pump.process_job(req_flow, duration)

print(pump.status())
```
