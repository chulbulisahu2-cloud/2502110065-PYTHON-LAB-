```python
import random

# Bin thresholds (Adjust to match your Job 09 specifications)
ACCEPT_MIN, ACCEPT_MAX = 24.95, 25.05

accept_count = 0
rework_count = 0
scrap_count = 0

batch_size = 10  # Set to 100 for the full bench run

print(f"--- Running Self-Sorting Bench ({batch_size} parts) ---")

for i in range(1, batch_size + 1):
    diameter = round(random.uniform(24.90, 25.10), 2)

    # Sorting logic
    if ACCEPT_MIN <= diameter <= ACCEPT_MAX:
        status = "ACCEPT"
        accept_count += 1
    elif 24.90 <= diameter < ACCEPT_MIN or ACCEPT_MAX < diameter <= 25.10:
        # Check if reworkable vs scrap based on tolerance bounds
        if 24.92 <= diameter <= 25.08:
            status = "REWORK"
            rework_count += 1
        else:
            status = "SCRAP"
            scrap_count += 1

    print(f"Part {i:03d} | Diameter: {diameter:.2f} mm | Status: {status}")

# Shift Report
print("\n=== SHIFT REPORT ===")
print(f"Total Inspected : {batch_size}")
print(f"Accepted        : {accept_count}")
print(f"Rework          : {rework_count}")
print(f"Scrap           : {scrap_count}")
```

```python
import random

ACCEPT_MIN, ACCEPT_MAX = 24.95, 25.05

mode = input("Select mode ('halt' to stop on scrap, 'skip' to log & continue): ").strip().lower()

accept_count = 0
rework_count = 0
scrap_count = 0
inspected = 0

batch_size = 50

print(f"\n--- Running in [{mode.upper()}] Mode ---")

for i in range(1, batch_size + 1):
    diameter = round(random.uniform(24.90, 25.10), 2)

    if ACCEPT_MIN <= diameter <= ACCEPT_MAX:
        status = "ACCEPT"
        accept_count += 1
    elif 24.92 <= diameter <= 25.08:
        status = "REWORK"
        rework_count += 1
    else:
        status = "SCRAP"
        scrap_count += 1

    inspected += 1

    if status == "SCRAP":
        if mode == "halt":
            print(f"Part {i:03d} | Diameter: {diameter:.2f} mm | Status: SCRAP --> HALTING BATCH!")
            break  # Stops the loop immediately
        elif mode == "skip":
            print(f"Part {i:03d} | Diameter: {diameter:.2f} mm | Status: SCRAP --> Logging & Skipping...")
            continue  # Skips further processing for this part and continues loop

    print(f"Part {i:03d} | Diameter: {diameter:.2f} mm | Status: {status}")

print("\n=== BATCH SUMMARY ===")
print(f"Total Inspected before termination: {inspected}")
print(f"Accepted: {accept_count} | Rework: {rework_count} | Scrap: {scrap_count}")
```

```python
import random

ACCEPT_MIN, ACCEPT_MAX = 24.95, 25.05

consecutive_accepts = 0
total_attempts = 0
MAX_ATTEMPTS = 500  # Industrial safety cap

print("--- Running Until 5 Consecutive ACCEPTs ---")

while consecutive_accepts < 5 and total_attempts < MAX_ATTEMPTS:
    total_attempts += 1
    diameter = round(random.uniform(24.90, 25.10), 2)

    if ACCEPT_MIN <= diameter <= ACCEPT_MAX:
        consecutive_accepts += 1
        status = f"ACCEPT (Streak: {consecutive_accepts})"
    else:
        consecutive_accepts = 0  # Reset counter on any non-ACCEPT part
        status = "REJECT/REWORK (Streak reset to 0)"

    print(f"Attempt {total_attempts:03d} | Diameter: {diameter:.2f} mm | {status}")

# Completion summary
if consecutive_accepts == 5:
    print(f"\nSUCCESS: Reached 5 consecutive ACCEPTs in {total_attempts} total machining attempts!")
else:
    print(f"\nSAFETY STOP: Reached safety limit of {MAX_ATTEMPTS} attempts without achieving 5 consecutive ACCEPTs.")
```
