```python
# 1. Store raw line and extract fields
raw_line = "  PE-26-0137 ; 25.02 ; DAY  "
fields = raw_line.strip().split(";")

part_code = fields[0].strip()
diameter = float(fields[1])  # Converts string to float
shift = fields[2].strip().upper()

# 2. Print labeled output and prove diameter is numeric
print(f"Part Code: {part_code}")
print(f"Diameter : {diameter}")
print(f"Shift    : {shift}")
print(f"Proof (+0.01): {diameter + 0.01}")
```

IT 10

```python
log_lines = [
    "  PE-26-0101 ; 25.02 ; DAY  ",
    "PE-26-0102 ; 24.95 ; DAY",
    "  PE-26-0103 ; 25.10 ; NIGHT  ",
    "PE-26-0104 ; 24.88 ; DAY ",
    " PE-26-0105 ; 25.00 ; NIGHT",
    "PE-26-0106 ; 25.05 ; DAY",
    "  PE-26-0107 ; 24.92 ; NIGHT ",
    "PE-26-0108 ; 25.01 ; DAY",
    "PE-26-0109 ; 24.99 ; NIGHT",
    " PE-26-0110 ; 25.08 ; DAY "
]

# Extract diameters into a list
# Choice: List retains insertion order and permits duplicates, unlike sets.
diameters = [float(line.split(";")[1]) for line in log_lines]  # list: maintains sequence and repeat values

# Summary stats
print(f"Count  : {len(diameters)}")
print(f"Average: {sum(diameters) / len(diameters):.2f}")
print(f"Min    : {min(diameters)}")
print(f"Max    : {max(diameters)}")
```

IT 8

```python
# Assuming tolerance bounds: Accepted [24.95, 25.05]
accepted_dias = []  # list: store accepted measurements in order
rework_dias = []    # list: store rework measurement values
scrap_dias = []     # list: store scrap measurement values
scrapped_codes = [] # list: store part codes flagged as scrap

for line in log_lines:
    parts = line.split(";")
    code = parts[0].strip()
    dia = float(parts[1])

    if 24.95 <= dia <= 25.05:
        accepted_dias.append(dia)
    elif 24.90 <= dia < 24.95 or 25.05 < dia <= 25.10:
        rework_dias.append(dia)
    else:
        scrap_dias.append(dia)
        scrapped_codes.append(code)

acc_avg = sum(accepted_dias) / len(accepted_dias) if accepted_dias else 0

print(f"Accepted Count : {len(accepted_dias)}")
print(f"Rework Count   : {len(rework_dias)}")
print(f"Scrap Count    : {len(scrap_dias)}")
print(f"Accepted Avg   : {acc_avg:.2f}")
print(f"Scrapped Parts : {scrapped_codes}")
```

IT 6

```python
day_scraps = set()   # set: distinct collection of scrapped codes for DAY shift
night_scraps = set() # set: distinct collection of scrapped codes for NIGHT shift

day_total, day_acc = 0, 0
night_total, night_acc = 0, 0

for line in log_lines:
    parts = line.split(";")
    code = parts[0].strip()
    dia = float(parts[1])
    shift = parts[2].strip().upper()

    is_accepted = 24.95 <= dia <= 25.05

    if shift == "DAY":
        day_total += 1
        if is_accepted:
            day_acc += 1
        elif dia < 24.90 or dia > 25.10:
            day_scraps.add(code)
    elif shift == "NIGHT":
        night_total += 1
        if is_accepted:
            night_acc += 1
        elif dia < 24.90 or dia > 25.10:
            night_scraps.add(code)

print(f"Day Accept Rate  : {(day_acc / day_total) * 100:.1f}%")
print(f"Night Accept Rate: {(night_acc / night_total) * 100:.1f}%")

# Set comparisons
both_shifts_scrap = day_scraps & night_scraps  # set intersection
unique_to_one_shift = day_scraps ^ night_scraps # symmetric difference

print(f"Scrapped in both shifts: {both_shifts_scrap}")
print(f"Scrapped in only one   : {unique_to_one_shift}")
```
