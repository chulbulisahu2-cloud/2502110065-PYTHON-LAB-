```python
# IT12: The Three Core Jigs

def inspect(dia, target=25.00, tolerance=0.05):
    """Classifies a part diameter into 'accept', 'rework', or 'scrap'."""
    if dia > (target + tolerance):
        return "rework"
    elif dia < (target - tolerance):
        return "scrap"
    else:
        return "accept"


def tally(labels):
    """Counts frequencies of bin labels and returns a dictionary of totals."""
    counts = {"accept": 0, "rework": 0, "scrap": 0}
    for label in labels:
        counts[label] = counts.get(label, 0) + 1
    return counts


def report(shift_name, counts):
    """Formats shift yield counts into a single printable report string."""
    acc = counts.get("accept", 0)
    rew = counts.get("rework", 0)
    scr = counts.get("scrap", 0)
    return f"{shift_name:<6} accept {acc} rework {rew} scrap {scr}"


# --- IT12 Testing ---
if __name__ == "__main__":
    print("Testing inspect():", inspect(25.02))  # accept
    print("Testing tally():", tally(["accept", "scrap", "accept"]))
    print("Testing report():", report("TEST", {"accept": 2, "rework": 0, "scrap": 1}))
```

```python
# IT8: Catalogue Lookup & Scrap Sets

def lookup(catalogue, part_code, default=0.0):
    """Safely retrieves a part diameter from the catalogue or returns default."""
    return catalogue.get(part_code, default)


def scrap_codes(catalogue, target=25.00, tolerance=0.05):
    """Returns a set of part codes whose dimensions fall outside specification."""
    scraps = set()
    upper = target + tolerance
    lower = target - tolerance

    for code, dia in catalogue.items():
        if dia > upper or dia < lower:
            scraps.add(code)

    return scraps


# --- IT8 Testing ---
parts_catalogue = {
    "P-101": 25.02,  # Accept
    "P-102": 24.90,  # Scrap (< 24.95)
    "P-103": 25.05,  # Accept
    "P-104": 25.09,  # Rework (> 25.05)
    "P-105": 24.80   # Scrap (< 24.95)
}

print("Lookup P-101:", lookup(parts_catalogue, "P-101"))
print("Lookup Missing P-999 (default):", lookup(parts_catalogue, "P-999", default=0.0))
print("Scrap/Rework Part Codes (Set):", scrap_codes(parts_catalogue))
```

```python
"""
shoptools.py - Production Engineering Tool Crib for Lab 06
"""

def inspect(dia, target=25.00, tolerance=0.05):
    """Classifies a part diameter into 'accept', 'rework', or 'scrap'."""
    if dia > (target + tolerance):
        return "rework"
    elif dia < (target - tolerance):
        return "scrap"
    else:
        return "accept"


def tally(labels):
    """Counts frequencies of bin labels and returns a dictionary of totals."""
    counts = {"accept": 0, "rework": 0, "scrap": 0}
    for label in labels:
        counts[label] = counts.get(label, 0) + 1
    return counts


def report(shift_name, counts):
    """Formats shift yield counts into a single printable report string."""
    acc = counts.get("accept", 0)
    rew = counts.get("rework", 0)
    scr = counts.get("scrap", 0)
    return f"{shift_name:<6} accept {acc} rework {rew} scrap {scr}"


def lookup(catalogue, part_code, default=0.0):
    """Safely retrieves a part diameter from the catalogue or returns default."""
    return catalogue.get(part_code, default)


def scrap_codes(catalogue, target=25.00, tolerance=0.05):
    """Returns a set of part codes whose dimensions fall outside tolerance."""
    upper = target + tolerance
    lower = target - tolerance
    return {code for code, dia in catalogue.items() if dia > upper or dia < lower}
```
